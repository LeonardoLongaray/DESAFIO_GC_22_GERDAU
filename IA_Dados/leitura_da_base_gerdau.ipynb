```python
import pandas as pd
import statistics as sts
```

```python
# Tentando ler o arquivo com o separador ponto e vírgula e pulando linhas inválidas, se houverem
df = pd.read_csv("../1_informacoes_do_desafio/base_de_dados_bruta.csv", sep=';', on_bad_lines='skip')

# Exibe as primeiras linhas para verificar se foi carregado corretamente
df.head(100)
```

#### **Ray:**
  Identifiquei duplicações ou mais que isso em alguns códigos na coluna Código do Serviço que estão logo abaixo:

```python
contagem_codigos = df["Código do Serviço"].value_counts()

contagem_codigos[contagem_codigos > 1]
```

```python
# Filtrando apenas as ocorrências de códigos duplicados (contagem > 1) com base na lógica da célula 8
contagem_servico = df["Código do Serviço"].value_counts()
duplicados_servico = contagem_servico[contagem_servico > 1]

# Calculando a média de ocorrências desses códigos duplicados
media_duplicados = duplicados_servico.mean()
print(f"A média de ocorrências por código duplicado é: {media_duplicados:.2f}")
```

#### **Ray:**
  Identifiquei duplicações ou mais que isso em alguns códigos na coluna Código do Fornecedor que estão logo abaixo (sem uso atual):

```python
contagem_codigos = df["Código do Fornecedor"].value_counts()

contagem_codigos[contagem_codigos > 1]
```

```python
# Calculando a moda da quantidade de repetições com a biblioteca statistics (sts)
moda_geral = sts.mode(contagem_servico)
moda_duplicados = sts.mode(duplicados_servico)

print(f"A moda estatística de repetições (todos os códigos) é: {moda_geral}")
print(f"A moda estatística de repetições (apenas códigos duplicados) é: {moda_duplicados}")
```
